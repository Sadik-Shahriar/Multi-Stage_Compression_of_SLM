# Notebook 09: Drive-Independent Cross-Model Synthesis

**CPU-only derived analysis.** This Colab notebook downloads the uploaded CSE407 repository ZIP with `gdown`, validates the immutable Llama 3.2-3B and Qwen2.5-3B evidence, and executes the complete registered cross-model synthesis without mounting Google Drive or reading `MyDrive`.

Inputs are extracted under `/content/cse407_uploaded_repo/`. The fail-closed 41-file result bundle and downloadable ZIP are written under `/content/`. Colab runtime storage is ephemeral, so download the final ZIP before disconnecting.

```
# Run 0: Acquire and verify the uploaded project archive
%pip -q install gdown

from pathlib import Path
import gdown
import shutil
import zipfile

DRIVE_FILE_ID = "1MtCVx8ZkSVVgQ6Q1SNNEOMnrqvh393O4"
DOWNLOAD_PATH = Path("/content/cse407_project.zip")
EXTRACT_ROOT = Path("/content/cse407_uploaded_repo")
LOCAL_OUTPUT_ROOT = Path("/content/CSE407_cross_model_synthesis_v1")

if LOCAL_OUTPUT_ROOT.exists():
    raise FileExistsError(
        f"Output already exists: {LOCAL_OUTPUT_ROOT}. "
        "The synthesis is fail-closed and will not overwrite it."
    )

if not DOWNLOAD_PATH.exists():
    result = gdown.download(id=DRIVE_FILE_ID, output=str(DOWNLOAD_PATH), quiet=False)
    if result is None:
        raise RuntimeError("gdown failed to download the uploaded repository archive")

if EXTRACT_ROOT.exists():
    shutil.rmtree(EXTRACT_ROOT)
EXTRACT_ROOT.mkdir(parents=True)
with zipfile.ZipFile(DOWNLOAD_PATH) as archive:
    unsafe = [n for n in archive.namelist() if Path(n).is_absolute() or ".." in Path(n).parts]
    if unsafe:
        raise RuntimeError(f"Unsafe ZIP members detected: {unsafe[:5]}")
    archive.extractall(EXTRACT_ROOT)

required_source = (
    "AGENTS.md",
    "research/experiment_registry.json",
    "research/llama_v5_artifact_manifest.json",
    "research/llama_track_a_full_v1_artifact_manifest.json",
    "research/llama_power_robustness_v1_artifact_manifest.json",
    "experiments/track_a_failsafe_colab/09_cross_model_synthesis_v1_colab.ipynb",
)
source_candidates = sorted({p.parent for p in EXTRACT_ROOT.rglob("AGENTS.md")})
source_matches = [p for p in source_candidates if all((p / rel).is_file() for rel in required_source)]
if len(source_matches) != 1:
    raise RuntimeError(f"Expected exactly one project root, found: {source_matches}")
SOURCE_ROOT = source_matches[0]

ARTIFACT_BASE = SOURCE_ROOT / "evidence" / "accepted" / "bundles"
required_bundles = (
    "CSE407_llama_v5_final",
    "CSE407_llama_track_a_full_v1",
    "CSE407_llama_power_robustness_v1",
    "CSE407_qwen_track_a_reference_v1",
)
missing = [name for name in required_bundles if not (ARTIFACT_BASE / name).is_dir()]
if missing:
    raise FileNotFoundError(f"Uploaded archive is missing accepted bundles: {missing}")

NOTEBOOK_PATH = SOURCE_ROOT / "experiments" / "track_a_failsafe_colab" / "09_cross_model_synthesis_v1_colab.ipynb"
print(f"Source root: {SOURCE_ROOT}")
print(f"Artifact base: {ARTIFACT_BASE}")
print(f"Notebook: {NOTEBOOK_PATH}")
print(f"Archive size: {DOWNLOAD_PATH.stat().st_size / 1e6:.1f} MB")
print("Preflight passed: registered source and four immutable bundles found; local output path is clean.")
```
